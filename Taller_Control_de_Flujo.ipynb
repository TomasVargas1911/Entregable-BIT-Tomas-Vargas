## 1. Registro de estudiantes
Creo la lista de estudiantes. Cada uno es una tupla con (nombre, edad, nota).

```python
estudiantes = [
    ("Camila", 22, 4.7),
    ("Andres", 20, 3.3),
    ("Miguel", 24, 2.8),
    ("Valentina", 19, 4.1),
    ("Pedro", 27, 3.9),
    ("Laura", 21, 5.0),
    ("Diego", 23, 4.5),
    ("Carolina", 26, 3.6),
    ("Felipe", 18, 2.4),
    ("Mariana", 25, 4.9),
]

print(estudiantes)
```

## 2. Mostrar la información
Recorro la lista con un for e imprimo los datos de cada estudiante.

```python
print("Registro de estudiantes:")
for nombre, edad, nota in estudiantes:
    print(f"{nombre} tiene {edad} años y obtuvo una nota de {nota}")
```

## 3. Clasificación de estudiantes
Uso if, elif y else para clasificar la nota de cada estudiante.

```python
print("\nClasificación de estudiantes:")
for nombre, edad, nota in estudiantes:
    if nota >= 4.5:
        clasificacion = "Excelente"
    elif nota >= 4.0:
        clasificacion = "Bueno"
    elif nota >= 3.0:
        clasificacion = "Aceptable"
    else:
        clasificacion = "Reprobó"
    print(f"{nombre} obtuvo {nota} y su clasificación es {clasificacion}")
```

## 4. Promedio general
Sumo todas las notas con una variable acumuladora y calculo el promedio.

```python
suma = 0
for nombre, edad, nota in estudiantes:
    suma = suma + nota

promedio = suma / len(estudiantes)
print(f"El promedio general de las notas es: {promedio:.2f}")
```

## 5. Búsqueda de estudiante
Pido un nombre por teclado y reviso si está en la lista.

```python
nombre_buscado = input("Escribe el nombre del estudiante que quieres buscar: ")

encontrado = False
for nombre, edad, nota in estudiantes:
    if nombre.lower() == nombre_buscado.lower():
        encontrado = True
        edad_encontrada = edad
        nota_encontrada = nota
        break

if encontrado:
    print(f"Sí, {nombre_buscado} está registrado. Tiene {edad_encontrada} años y su nota es {nota_encontrada}")
else:
    print(f"No encontré a ningún estudiante llamado {nombre_buscado}")
```

## 6. Diccionario de ciudades
Creo un diccionario donde la llave es el nombre del estudiante y el valor es su ciudad.

```python
estudiante_ciudad = {
    "Camila": "Bogota",
    "Andres": "Medellin",
    "Miguel": "Cali",
    "Valentina": "Bogota",
    "Pedro": "Cartagena",
    "Laura": "Medellin",
    "Diego": "Bogota",
    "Carolina": "Cali",
    "Felipe": "Cartagena",
    "Mariana": "Medellin",
}

for nombre, ciudad in estudiante_ciudad.items():
    print(f"{nombre} vive en {ciudad}")
```

## 7. Cantidad de estudiantes por ciudad
A partir del diccionario anterior armo uno nuevo que cuenta cuántos estudiantes hay en cada ciudad.

```python
conteo_por_ciudad = {}

for ciudad in estudiante_ciudad.values():
    if ciudad not in conteo_por_ciudad:
        conteo_por_ciudad[ciudad] = 1
    else:
        conteo_por_ciudad[ciudad] += 1

print(conteo_por_ciudad)
```

## 8. Ciclo while
Pido números hasta que el usuario escriba 0. Al final muestro cuántos ingresó, la suma total y el promedio.

```python
suma_total = 0
cantidad = 0

while True:
    numero = float(input("Ingresa un número (0 para terminar): "))
    if numero == 0:
        break
    suma_total += numero
    cantidad += 1

if cantidad > 0:
    print(f"Cantidad de números ingresados: {cantidad}")
    print(f"Suma total: {suma_total}")
    print(f"Promedio: {suma_total / cantidad:.2f}")
else:
    print("No ingresaste ningún número antes del 0")
```

## 9. Uso de break y continue
Recorro los números del 1 al 30, salto los múltiplos de 3 con continue y me detengo cuando llego a 25 con break.

```python
numeros_procesados = []

for i in range(1, 31):
    if i == 25:
        break
    if i % 3 == 0:
        continue
    numeros_procesados.append(i)

print("Números procesados:")
print(numeros_procesados)
```
